```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    "StudyHours": [1,2,3,4],
    "PreviousMarks": [2,3,5,4]
}
df = pd.DataFrame(data)
print("Original Dataset:")
print(df)
```

```python
scaler = StandardScaler()
standardized_data = scaler.fit_transform(df)

standardized_df = pd.DataFrame(
    standardized_data,
    columns = df.columns
)
print("\nStandardized Dataset:")
print(standardized_df.round(6))
```

```python
covariance_matrix = np.cov(
    standardized_data,
    rowvar=False
)
print("\nCovariance Matrix:")
print(covariance_matrix.round(6))
```

```python
eigenvalues, eigenvector = np.linalg.eig(
    covariance_matrix

)
eigenvalues = eigenvalues.real
eigenvector = eigenvector.real
print("\nEigenvalues:")
print(eigenvalues.round(6))

print("Eigenvectors:")
print(eigenvector.round(6))
```

```python
index = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[index]
eigenvector = eigenvector[:, index]

print("\nSorted Eigenvalues:")
print(eigenvalues,round(6))

print("\n Sorted Eigenvectors:")
print(eigenvector.round(6))
```

```python
explained_variance = (
    eigenvalues / eigenvalues.sum())
print("\n Explained Variance Ratio:")
print(explained_variance.round(6))
```

```python
cumulative_variance = np.cumsum(explained_variance)

print("\n Cumulative Explained Variance")
print(cumulative_variance.round(6))
```

```python
principal_components = np.dot(standardized_data,eigenvector)
pca_df = pd.DataFrame(
    principal_components,
    columns=["PC1","PC2"]
)


print("\n Principal Components:")
print(pca_df.round(6))
```

```python
reduced_data=principal_components[:,0]
reduced_data=pd.DataFrame(reduced_data, columns=["PC1"])

print("\nReduced Data:")
print(reduced_data.round(6))
```
