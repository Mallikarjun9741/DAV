```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("students.csv")
print(df)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(
    df["Marks"],
    bins=10


)
```

```python
plt.figure(figsize=(8, 4))
sns.boxplot(
    x=df["Marks"]
)
plt.title("Box plot os student Marks")
plt.xlabel("Marks")
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x="StudyHours",
    y="Marks",
    data=df,
    hue="Result",
    palette={
        "Pass": "green",
        "Fail": "red"
    },
    s=70
)

# Add overall trend line
sns.regplot(
    x="StudyHours",
    y="Marks",
    data=df,
    scatter=False,
    color="blue"
)

plt.title("Study Hours vs Marks by Result")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.legend(title="Result")
```

```python
sns.pairplot(
    df[
        [
            "StudyHours",
            "Attendance",
            "PrevMarks",
            "Marks"
        ]
        ]
)
```

```python
from numpy import correlate
correlation = df[
    [
        "StudyHours",
        "Attendance",
        "PrevMarks",
        "Marks"
    ]
].corr()
print("\n Correlation Marks:")
print(correlation)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True
)
plt.title("Correlation HeatMap")
plt.show()
```
