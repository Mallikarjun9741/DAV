```python
new_sales = pd.DataFrame({
    "Order Date": [
        "2024-06-01",
        "2024-06-05",
        "2024-06-10"
    ],
    "Product Name":[
        "Laptop",
        "Mouse",
        "Printer"
    ],
    "Category": [
        "Electronics",
        "Accessories",
        "Office"
    ],

    "Region": [
        "North",
        "South",
        "East"
    ],

    "Quantity":[
        2,
        4,
        3
    ],
    "Sales": [
        60000,
        2000,
        9000
    ],

    "Profit": [
        12000,
        400,
        1800
    ],

})
print("11. NEW SALES DATAFRAME")
print("_"*50)
display(new_sales)
```

```python
combined_df = pd.concat(
    [df,new_sales],
    ignore_index=True
)
```

```python
print("12. CONCATENATED DATAFRAME")
print("_"*50)
display(combined_df)
```

```python
product_info = pd.DataFrame({
    "Product Name": [
        "Laptop",
        "Mouse",
        "printer",
        "Tablet",
        "Monitor"
    ],
    "supplier":[
        "Dell",
        "Logitech",
        "HP",
        "Samsung",
        "LG"

    ],
    "Discount": [
        10,
        5,
        15,
        8,
        12
    ]

})

print("13. PRODUCT INFORMATION DATAFRAME")
print("_"*50)
display(product_info)
```

```python
merged_df = pd.merge(
    df,
    product_info,
    on="Product Name",
    how="left"
)
print("14. MERGED DATAFRAME")
print("_"*50)
display(merged_df)
```

```python
region_info = pd.DataFrame(
    {
    "Region_Manager": [
    "Manager A",
    "Manager B",
    "Manager C",
    "Manager D"
    ]
    },
    index=[
        "North",
        "South",
        "East",
        "West"

    ]
)
print("15.  REGION INFORMATION DATAFRAME")
print("_"*50)
display(region_info)
```

```python
df_region=df.set_index("Region")
joined_df = df_region.join(region_info)
print("16. JOINED DATAFRAME")
print("_"*50)
display(joined_df)
```

```python
region_sales = df.groupby(
    "Region"
)["Sales"].sum()
print("17. TOTAL SALES BY REGION " )
print("_"*50)
display(region_sales)
```

```python
pivot_table = pd.pivot_table(
    df,
    values="Sales",
    index="Category",
    columns="Region",
    aggfunc="sum",
    fill_value=0
)
print("18  PIVOT TABLE.")
print("_"*50)
display(pivot_table)
```
