```python
import numpy as np
```

```python
import pandas as pd
```

```python
arr=np.array([10,20,30,40,50])
print("Numpy Array:")
print(arr)
```

```python
print("---Array Indexing -----")
print("First element:",arr[0])
print("Third element:",arr[2])
print("Last element:",arr[-1])
```

```python
print("First Three elements :",arr[:3])
print("Element from index 2:",arr[2:])
print("Last three element:",arr[-3:])
```

```python
print("\n----Basic Array Operation")
print("Addition :",arr+5)
print("Substaction :",arr-5)
print("Multiplication by 2:",arr*2)
print("Divided by 10 :",arr/10)
```

```python
print("\n----Mathematical Operations---")
print("Sum: ",np.sum(arr))
print("Mean:",np.mean(arr))
print("Maxium:",np.max(arr))
print("Minium:",np.min(arr))
```

```python
print("\n----Array Reshape----")
print("Reshape array")
reshaped_arr=arr.reshape(5,1)
print(reshaped_arr)
```

```python
import pandas as pd
```

```python
data ={
    "Name": ["Asha","Raghav","Ravi","Meera","Karan"],
    "Marks": [88,96,85,65,90],
    "Brach": ["CSE","ECE","ISE","MEC","CSE"]
}
```

```python
df = pd.DataFrame(data)
print("\n---Pandas DataFrame ---")
print(df)
```

```python
print("\n---Column Indexing----")
print("Name coloum:")
print(df["Marks"])
```

```python
print("\n----Row Indexing----")
print("First two rows:")
print(df.iloc[0:2])
```

```python
print("\n---Specific value---")
print("Marks of first student :",df.iloc[0,1])
```

```python
print("\n---Rows and column slicing----")
print("First three rows with  Name and Marks:")
print(df.iloc[0:3,[0,1]])
```

```python
print("\n----Dta filtering----")
result=df[df["Marks"] > 70]
print("Students with Marks greter than 70:")
print(result)
```
